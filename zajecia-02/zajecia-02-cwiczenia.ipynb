### B1. Kod osoby badanej

Kod osoby ma postać `P07-K-23`: numer osoby, płeć (`K` albo `M`) i wiek. Napisz program, który wytnie z kodu numer, płeć i wiek, a wiek zamieni na liczbę (`int`). Żeby sprawdzić, że wiek jest liczbą, wypisz też wiek za rok.

````python
kod = "P07-K-23"

# Twój kod
numer = kod[0:3]
plec = kod[4]
wiek = int(kod[6:])
wiek_za_rok = wiek + 1
print(f"numer: {numer}")
print(f"płeć: {plec}")
print(f"wiek: {wiek}")
print(f"wiek za rok: {wiek_za_rok}")
````

**Oczekiwany wynik:**

```text

płeć: K
wiek: 23
wiek za rok: 24
```

### B2. Zgoda z formularza

````python
odpowiedz = "  TAK "

# Twój kod
odpowiedz = odpowiedz.strip().lower()
zgoda = odpowiedz == "tak"
print(odpowiedz, zgoda)
````
